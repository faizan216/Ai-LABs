**Maximum_Repeating_Word**

```python
def max_repeating(text):

    count = {}

    for word in text.split():
        count[word] = count.get(word, 0) + 1

    return max(count, key=count.get)


tex = "My my my faizan"

m = max_repeating(tex)

m
```

```python
import functools

def greetig(fn):
    @functools.wraps(fn)
    def function(*args, **kwargs):
        print("Faizan")
        result = fn(*args, **kwargs)
        print("Asif")
        return result
    return function

@greetig
def su(a, b):
    return a+b

su(1,2)
```
