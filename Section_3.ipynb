**Creating Comprehesion List**

```python
list = [x*x*x for x in range(1,5)]
print(list)
```

**Separating Vowels**

```python
list = [x for x in "artificial intelligence" if x == 'a' or x == 'A' or x == 'E' or x == 'e' or x == 'I' or x == 'i' or x == 'O' or x == 'o' or x == 'U' or x == 'u' ]
print(list)
```

**Even Odd in Dictionary**

```python
dic = {x: ("Even" if x % 2 == 0 else "Odd") for x in range(10)}
dic
```
