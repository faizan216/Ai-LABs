**Check Integer Type**

```python
def classify(n):
    if n > 0:
        return "Positve"
    elif n < 0:
        return "Negative"
    else: return "zero"

n = classify(32)
print(n)

m = int(input("Enter:"))
result = "postive" if m > 0 else("Neg" if m < 0 else "zero")
print(result)
```

**Check palindrome**

```python
def check_palindrome(string):
    return string == string[::-1]

text = input("Enter String:")
output = check_palindrome(text)
print(output)
```

**Sum of multiples of 3 & 5**

```python
def sum_of_three_five(n):
    count = 0
    for i in range(n):
        if i % 3 == 0 or i % 5 == 0:
            count += i
    return count

n1 = sum_of_three_five(20)
n1
```
