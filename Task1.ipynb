```python
def filter_and_sor(numbers, threshold):
    num =[]
    i = 0
    n = len(numbers)
    for i in range(n):
        if numbers[i] > threshold:
            num.append(numbers[i])
    return sorted(num)

li = filter_and_sor([4,15,2,9,21,6],5)
print(li)
```

```python
def diamond(n):
    m = n // 2 + 1          # number of rows in top half (including middle)

    # top half: 1, 3, 5, ... up to n stars
    for i in range(m):
        stars = 2 * i + 1
        spaces = (n - stars) // 2
        print(" " * spaces + "*" * stars)

    # bottom half: mirror, skipping the middle row already printed
    for i in range(m - 2, -1, -1):
        stars = 2 * i + 1
        spaces = (n - stars) // 2
        print(" " * spaces + "*" * stars)

diamond(5)
```
